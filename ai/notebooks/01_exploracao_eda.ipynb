# 01 — Exploracao de Dados (EDA)

**Vitalis PI** — Dataset Health & Lifestyle preprocessado.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('../data/dataset_preprocessado.csv')
df.head()
```

```python
df.info()
df.describe()
```

```python
sns.countplot(data=df, x='Health_Profile', order=df['Health_Profile'].value_counts().index)
plt.title('Distribuicao de Perfis')
plt.xticks(rotation=15)
plt.show()
```

```python
numeric = ['Age','BMI','Daily_Steps','Hours_of_Sleep','Exercise_Hours_per_Week']
sns.heatmap(df[numeric].corr(), annot=True, cmap='coolwarm')
plt.show()
```
