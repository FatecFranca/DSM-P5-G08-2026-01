# 04 — Clusterizacao (K-Means)

Aprendizado **nao supervisionado**.

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

FEATURE_COLS = ['Age','BMI','Daily_Steps','Hours_of_Sleep','Exercise_Hours_per_Week','Heart_Rate','Calories_Intake','Alcohol_Consumption_per_Week','Gender_enc','Smoker_enc','Diabetic_enc','Heart_Disease_enc']

df = pd.read_csv('../data/dataset_preprocessado.csv')
X = df[FEATURE_COLS]
km = KMeans(n_clusters=4, random_state=42, n_init=10).fit(X)
print('Silhouette:', silhouette_score(X, km.labels_))
print('Clusters:', pd.Series(km.labels_).value_counts().sort_index())
```
